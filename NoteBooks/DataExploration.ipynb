```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
allPrices = yf.download(tickers = "SPY", interval="1d", period="10y", auto_adjust=True)
allPrices
```

```python
closingPrices = allPrices["Close"]
closingPrices
```

Using Adjusted Close Price to account for stock prices changing after closing for example when a company pays out dividends

```python
closingPrices.plot()
```

Getting Returns
return = final_price/initial_price - 1

to calculate multi period returns = (1 + return1)(1+return2) - 1
so we can get the compound returns for ETF over the 10y period

```python
returns = closingPrices.pct_change().dropna()
returns
```

```python
returns.plot()
```

```python
compoundReturns = (returns + 1).prod() - 1
(compoundReturns*100).round(2).astype("str") + "%"
```

```python
print("Largest return is:", returns.max())
print("Average return is:", returns.mean())
print("Standard Deviation is:", returns.std())
```

Rolling Volatilly Graph (20-days)

It is the standard Deviation calulated using the previous 19 days and itself
There is 20 trading days in a month
showes how the volatilly changes over time


```python
rollingVolatilly = returns.rolling(window=20).std().dropna()
# first 19 days are blank
rollingVolatilly
```
