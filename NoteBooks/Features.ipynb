```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
allPrices = yf.download(tickers="SPY", interval="1d", start="2015-11-01", end="2026-01-01", auto_adjust=True)
allPrices
```

```python
returns = allPrices["Close"].pct_change().dropna()
returns
```

Going to set up the data for the machine learning

Going to use: 

5-day Return
20-day Return
20-day Volatilly
Volume Ratio

Making a new dataframe to hold the information
The target is if tommorrow is up or not

```python
fiveDayReturn = allPrices["Close"].pct_change(5).dropna()
twentyDayReturn = allPrices["Close"].pct_change(20).dropna()
twentyDayVolatilty = returns.rolling(window=20).std().dropna()
VolumeRatio = (allPrices["Volume"]/allPrices["Volume"].rolling(window=20).mean()).dropna()
TommorrowUp = allPrices["Close"].shift(-1) >  allPrices["Close"]
```

```python
data = pd.concat(
    {
        "5-Day Returns": fiveDayReturn,
        "20-Day Returns": twentyDayReturn,
        "20-Day Volatility": twentyDayVolatilty,
        "Volume Ratio": VolumeRatio,
        "Tommorrow Up": TommorrowUp
    },
    axis=1,
    sort=True
)

data = data.loc["2016-01-01":]
data.columns = data.columns.droplevel("Ticker")
data = data.dropna()
data
```

```python
data.to_csv("C:/Users/Miles/Documents/ml_project_cv/Data/features.csv")
trainingSet = data.loc["2016-01-01":"2023-01-01"]
validationSet = data.loc["2023-01-01":"2024-01-01"]
testSet = data.loc["2024-01-01":"2026-01-01"]
```

```python
trainingSet.to_csv("C:/Users/Miles/Documents/ml_project_cv/Data/trainSet.csv")
validationSet.to_csv("C:/Users/Miles/Documents/ml_project_cv/Data/validationSet.csv")
testSet.to_csv("C:/Users/Miles/Documents/ml_project_cv/Data/testSet.csv")
```
