```python
trainSet = pd.read_csv("C:/Users/Miles/Documents/ml_project_cv/Data/trainSet.csv", index_col=0, parse_dates=True)
validationSet = pd.read_csv("C:/Users/Miles/Documents/ml_project_cv/Data/validationSet.csv", index_col=0, parse_dates=True)
testSet = pd.read_csv("C:/Users/Miles/Documents/ml_project_cv/Data/testSet.csv", index_col=0, parse_dates=True)
label = "Tommorrow Up"
```

```python
allTrees = {}
maxSize = 5
allQuestions = CreateSetsOfQuestions(trainSet, label)
for size in range(1,maxSize):
    startNode = TreeBuilder(trainSet,size,allQuestions, label)
    allTrees[size] = startNode
```

```python
treeNumber = 0
bestCorrect = 0
validationResult = None
for size in range(1,maxSize):
    results, correct = Testing(allTrees[size], validationSet, label)
    if correct > bestCorrect:
        treeNumber = size
        validationResult = results

validationResult =  validationResult[["Prediction"]]

print("")
PrintTree(allTrees[treeNumber])
```

```python

decisionTreeResult, correct = Testing(allTrees[treeNumber], testSet, label)

decisionTreeResult = decisionTreeResult[["Prediction"]]

decisionTreeResult.to_csv("C:/Users/Miles/Documents/ml_project_cv/Predictions/DecisionTree.csv")
```

```python
noOfTrees = [5,10]
maxDepth = [1,2]

Forests = {}

for noOfTree in noOfTrees:
    for depth in maxDepth:
        Forests[str(noOfTree)+str(depth)] = RandomForestBuilder(trainSet, label, noOfTree, depth)
```

```python



for noOfTree in noOfTrees:
    for depth in maxDepth:
        try:
            result, correct = FinalTest(Forests[str(noOfTree)+str(depth)], validationSet, label)

            if correct > maxCorrect:
                bestForest = str(noOfTree)+str(depth)
                maxCorrect = correct
        except:
            pass


randomForestResult, correct = FinalTest(Forests[bestForest], testSet, label)

randomForestResult = randomForestResult[["Prediction"]]

randomForestResult.to_csv("C:/Users/Miles/Documents/ml_project_cv/Predictions/RandomForest.csv")
randomForestResult

```
